```python
'''Task 2.4: This code fixes a specific data entry error:
a single physical store having contradictory sizes.
It ensures that every row for a specific store shares the exact
same size by finding the most frequently 
recorded size for that store and applying it across the board.'''
df['Outlet Size'] = df.groupby('Outlet Identifier')['Outlet Size'].transform(
    lambda s: s.mode().iloc[0] if not s.mode().empty else np.nan
)
```

```python
fig, axes = plt.subplots(1, len(numeric_cols), figsize=(15, 5))
for i, col in enumerate(numeric_cols):
    sns.boxplot(y=df[col], ax=axes[i]).set_title(col)
plt.tight_layout()
plt.show()
```

```python
# Task 2.3: Find values that look unusually high or unusually low (Outliers)
def count_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    return ((series < (Q1 - 1.5 * IQR)) | (series > (Q3 + 1.5 * IQR))).sum()

print("Outlier Counts (Kept in dataset):")
for col in numeric_cols:
    print(f"{col}: {count_outliers(df[col])}")
```

```python
# Task 3.4 – Better vs lower performers
by_type = df.groupby('Item Type')['Sales'].sum().sort_values()
print('Lowest 3:'); print(by_type.head(3).round(0))
print('Highest 3:'); print(by_type.tail(3).round(0))
```

```python
# Compare typical rating, visibility and weight of the 3 best vs 3 worst item types
best = by_type.tail(3).index
worst = by_type.head(3).index
cols = ['Sales', 'Rating', 'Item Visibility', 'Item Weight']
compare = pd.DataFrame({
    'Best 3 types': df[df['Item Type'].isin(best)][cols].mean(),
    'Worst 3 types': df[df['Item Type'].isin(worst)][cols].mean()})
compare.round(3)
```

```python
compare.loc[['Rating', 'Item Weight']].plot.bar(rot=0)
plt.title('Best vs worst item types')
plt.show()
```

```python
# Number of sales records and average sale for each outlet type
outlet_stats = df.groupby('Outlet Type').agg(records=('Sales', 'size'), average_sale=('Sales', 'mean'), total_sales=('Sales', 'sum'))
outlet_stats.round(1)
```
